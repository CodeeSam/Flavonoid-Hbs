# Scaffold-held-out checkpoint evaluation

Evaluate the archived epoch-25 checkpoint with the original training tokenizer and batch-averaging procedure. No model training or parameter updates are performed.
The shared evaluator first reproduces the saved validation loss, then evaluates the held-out test set. It uses PyTorch when available, otherwise the independent NumPy inference implementation.


```python
from pathlib import Path
import subprocess, sys
start = Path.cwd().resolve()
root = next((p for p in [start, *start.parents] if (p / "models").is_dir() and (p / "data").is_dir()), None)
if root is None:
    raise FileNotFoundError("Run this notebook from the repository root or notebooks directory.")
subprocess.run([sys.executable, str(root / "scripts/evaluate_heldout_test.py"), "--backend", "auto"], check=True)
```
